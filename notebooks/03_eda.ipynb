```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned data
df = pd.read_csv('../data/cleaned_data.csv')

# Set plot style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

print(f"Shape: {df.shape}")
print("Data loaded successfully")
```

```python
# Plot 1: Readmission rate by age group
age_readmission = df.groupby('age')['target'].mean() * 100

plt.figure(figsize=(10, 6))
sns.barplot(x=age_readmission.index, y=age_readmission.values, color='steelblue')
plt.title('Readmission rate by age group', fontsize=14)
plt.xlabel('Age (midpoint)')
plt.ylabel('Readmission rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../data/readmission_by_age.png')
plt.show()
print("Chart saved!")
```

```python
# Plot 2: Readmission rate by number of medications
plt.figure(figsize=(10, 6))
med_readmission = df.groupby('num_medications')['target'].mean() * 100
sns.lineplot(x=med_readmission.index, y=med_readmission.values, color='steelblue')
plt.title('Readmission rate by number of medications', fontsize=14)
plt.xlabel('Number of medications')
plt.ylabel('Readmission rate (%)')
plt.tight_layout()
plt.savefig('../data/readmission_by_medications.png')
plt.show()
print("Chart saved!")
```

```python
# Plot 3: Readmission rate by number of inpatient visits
plt.figure(figsize=(10, 6))
inpatient_readmission = df.groupby('number_inpatient')['target'].mean() * 100
sns.barplot(x=inpatient_readmission.index, y=inpatient_readmission.values, color='steelblue')
plt.title('Readmission rate by prior inpatient visits', fontsize=14)
plt.xlabel('Number of prior inpatient visits')
plt.ylabel('Readmission rate (%)')
plt.tight_layout()
plt.savefig('../data/readmission_by_inpatient.png')
plt.show()
print("Chart saved!")
```

```python
# Plot 4: Readmission rate by diagnosis group
plt.figure(figsize=(12, 6))
diag_readmission = df.groupby('diag_1')['target'].mean() * 100
diag_readmission = diag_readmission.sort_values(ascending=False).head(15)
sns.barplot(x=diag_readmission.values, y=diag_readmission.index, color='steelblue')
plt.title('Top 15 diagnoses by readmission rate', fontsize=14)
plt.xlabel('Readmission rate (%)')
plt.ylabel('Primary diagnosis code')
plt.tight_layout()
plt.savefig('../data/readmission_by_diagnosis.png')
plt.show()
print("Chart saved!")
```

```python
# Summary of key EDA findings
print("=== Key EDA Findings ===")
print(f"\n1. Highest readmission age group: 85 years")
print(f"\n2. Prior inpatient visits strongly predict readmission")
print(f"\n3. More medications = higher readmission risk (up to ~30 meds)")
print(f"\n4. Social factors (housing, nutrition) are top diagnosis predictors")

# Save final EDA dataset
df.to_csv('../data/eda_complete.csv', index=False)
print("\nEDA complete. Data saved.")
```
